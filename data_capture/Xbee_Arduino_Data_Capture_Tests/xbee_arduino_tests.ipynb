```python
import pandas as pd
import numpy as np

import matplotlib
%matplotlib inline
# matplotlib.use('TkAgg')  # or 'QtAgg', etc.

import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D  # Required for 3D plotting
```

```python
# Load tab-delimited file
# df = pd.read_csv('moving_in_circles.txt', delimiter='\t', header=None)
df = pd.read_csv('moving_up_and_down.txt', delimiter='\t', header=None)
df.columns = ['timestamp_ms', 'x', 'y', 'z']

# Extract as NumPy arrays
t = (1/1000)*df['timestamp_ms'].to_numpy()
x = df['x'].to_numpy()
y = df['y'].to_numpy()
z = df['z'].to_numpy()

print(f'Data array length = {len(t)}')
```

```python
# Test plotting of position data
plt.figure()
plt.plot(t, x, label='x')
plt.plot(t,y, label='y')
plt.plot(t,z, label='z')
plt.xlabel('Time (s)')
plt.ylabel('Position (m)')
plt.title("Position data")
plt.legend()
plt.show()
```

```python
# Calculating frequency at which data is received
est_freq = 1 / np.mean(np.diff(t))  # in Hz
# Printing estimated frequency
print(f"Estimated Frequency: {est_freq} Hz")
```

```python
# Test 3D trajectory plot

# Create a 3D figure and axis
fig = plt.figure()
ax = plt.axes(projection='3d')

[t_start_idx, t_end_idx] = [5000,8000]

# Plot the trajectory
ax.plot(x[t_start_idx:t_end_idx], y[t_start_idx:t_end_idx], z[t_start_idx:t_end_idx], label='Trajectory', color='blue')

# Optionally: highlight start and end points
ax.scatter(x[t_start_idx], y[t_start_idx], z[t_start_idx], color='green', label='Start', s=50)
ax.scatter(x[t_end_idx], y[t_end_idx], z[t_end_idx], color='red', label='End', s=50)

# Label axes
ax.set_xlabel('X Position')
ax.set_ylabel('Y Position')
ax.set_zlabel('Z Position')
ax.set_title('3D Trajectory')
ax.legend()

# Show the plot
plt.show()
```

# Checksum Testing

```python
# Load tab-delimited file
# df = pd.read_csv('moving_in_circles.txt', delimiter='\t', header=None)
df = pd.read_csv('position_checksum_test_1.txt', delimiter='\t', header=None)
df.columns = ['timestamp_ms', 'x', 'y', 'z', 'valid']

# Extract as NumPy arrays
t = (1/1000) * df['timestamp_ms'].to_numpy()
x = df['x'].to_numpy()
y = df['y'].to_numpy()
z = df['z'].to_numpy()
valid = df['valid'].to_numpy()

print(f'Data array length = {len(t)}')

# Test plotting of position data
plt.figure()
plt.plot(t, x, label='x')
plt.plot(t,y, label='y')
plt.plot(t,z, label='z')
plt.xlabel('Time (s)')
plt.ylabel('Position (m)')
plt.title("Position data")
plt.legend()
plt.show()

# Calculating frequency at which data is received
est_freq = 1 / np.mean(np.diff(t))  # in Hz
# Printing estimated frequency
print(f"Estimated Frequency: {est_freq} Hz")

# Test 3D trajectory plot

# Create a 3D figure and axis
fig = plt.figure()
ax = plt.axes(projection='3d')

[t_start_idx, t_end_idx] = [0,5000]

# Plot the trajectory
ax.plot(x[t_start_idx:t_end_idx], y[t_start_idx:t_end_idx], z[t_start_idx:t_end_idx], label='Trajectory', color='blue')

# Optionally: highlight start and end points
ax.scatter(x[t_start_idx], y[t_start_idx], z[t_start_idx], color='green', label='Start', s=50)
ax.scatter(x[t_end_idx], y[t_end_idx], z[t_end_idx], color='red', label='End', s=50)

# Label axes
ax.set_xlabel('X Position')
ax.set_ylabel('Y Position')
ax.set_zlabel('Z Position')
ax.set_title('3D Trajectory')
ax.legend()

# Show the plot
plt.show()

# Plot validity of data
plt.figure()
plt.plot(t, valid, label='validity')
plt.title("Validity of data over time")
plt.xlabel("Time (s)")
plt.ylabel("Validity")
plt.legend()
plt.show()

# Plot validity and data together in subplots
plt.figure()
plt.subplot(4,1,1)
plt.plot(t, x, label='x')
plt.ylabel('X (m)')
plt.legend()
plt.subplot(4,1,2)
plt.plot(t, y, label='y')
plt.ylabel('Y (m)')
plt.legend()
plt.subplot(4,1,3)
plt.plot(t, z, label='z')
plt.ylabel('Z (m)')
plt.legend()
plt.subplot(4,1,4)
plt.plot(t, valid, label='validity', color='orange')
plt.ylabel('Validity')
plt.xlabel("Time (s)")
plt.legend()
plt.suptitle("Position data and validity over time")
plt.show()
```
